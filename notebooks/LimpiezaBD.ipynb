```python
import pandas as pd

url = "https://raw.githubusercontent.com/AngyArango-G2/Reto-Tecnologico-Grupo2/refs/heads/main/data/raw/Numero_de_Pasajeros_que_salieron_desde_las_Terminales_de_Transporte_de_Medell%C3%ADn_en_Junio_a%C3%B1o_2019_20260416.csv"

df2 = pd.read_csv(url)

df2.head()
```

```python
df2.info()
```

```python
# Copia base
df2_clean = df2.copy()

# Eliminar duplicados
df2_clean = df2_clean.drop_duplicates()

# Renombrar columnas
df2_clean.columns = df2_clean.columns.str.strip().str.upper()

# Limpiar textos
for col in df2_clean.select_dtypes(include='object').columns:
    df2_clean[col] = df2_clean[col].astype(str).str.strip().str.upper()

print(df2_clean.shape)
```

```python
df2_clean["FECHA_PROGRAMADA"] = pd.to_datetime(df2_clean["FECHA_PROGRAMADA"], errors="coerce")
df2_clean["FECHASALIDA"] = pd.to_datetime(df2_clean["FECHASALIDA"], errors="coerce")

df2_clean["HORA_PROGRAMADA"] = pd.to_datetime(df2_clean["HORA_PROGRAMADA"], errors="coerce").dt.time
df2_clean["HORASALIDA"] = pd.to_datetime(df2_clean["HORASALIDA"], errors="coerce").dt.time
```

```python
import pandas as pd

df2_clean = pd.read_csv("/content/drive/MyDrive/RetoEafit/pasajeros_terminales_limpia.csv")
df3_clean = pd.read_csv("/content/drive/MyDrive/RetoEafit/velocidad_trafico_limpia.csv")
df4_clean = pd.read_csv("/content/drive/MyDrive/RetoEafit/camaras_cctv_limpia.csv")
df5_clean = pd.read_csv("/content/drive/MyDrive/RetoEafit/paradas_limpia.csv")
df6_clean = pd.read_csv("/content/drive/MyDrive/RetoEafit/rutas_limpia.csv")
df7_clean = pd.read_csv("/content/drive/MyDrive/RetoEafit/intensidad_vehicular_limpia.csv")

print("Bases cargadas correctamente")
```

```python
bases = {
    "pasajeros": df2_clean,
    "trafico": df3_clean,
    "camaras": df4_clean,
    "paradas": df5_clean,
    "rutas": df6_clean,
    "intensidad": df7_clean
}
```

```python
for nombre, df in bases.items():
    print(f"\n===== {nombre.upper()} =====")
    columnas_texto = df.select_dtypes(include='object').columns

    for col in columnas_texto:
        if df[col].nunique() <= 15:
            print(f"\n📌 {col}")
            print(df[col].value_counts())
```

```python
# ---------- TRAFICO ----------
df3_clean["NOMBRE_DÍA"] = df3_clean["NOMBRE_DÍA"].replace({
    "MI�RCOLES": "MIERCOLES",
    "S�BADO": "SABADO"
})

# ---------- INTENSIDAD ----------
df7_clean["OPERACION"] = df7_clean["OPERACION"].replace("NAN", "SIN DATO")

df7_clean["SENTIDO"] = df7_clean["SENTIDO"].replace({
    "0": "SIN DATO",
    "WE-EW": "WE",
    "NS-SN": "NS"
})

# limpiar comunas raras
valores_invalidos = [
    "SISTEMA VIAL DEL RI",
    "CORREGIMIENTO DE SAN CRISTOBAL"
]

df7_clean["NOMBRE_COMUNA"] = df7_clean["NOMBRE_COMUNA"].replace(valores_invalidos, "SIN DATO")
```
