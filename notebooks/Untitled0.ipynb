```python
import pandas as pd

# 1. Cargar el archivo con el separador y tipo de dato correcto
ruta_archivo = '/content/drive/My Drive/velocidad_trafico_limpia_final.csv'
df = pd.read_csv(ruta_archivo, sep=';', dtype={'id': str})

# 2. Limpiar nombres de columnas (quita espacios invisibles)
df.columns = df.columns.str.strip()

# 3. Validación de Identificadores (Paso técnico esencial)
df = df.dropna(subset=['id', 'fecha_hora'])

# 4. Ajuste de Formatos (Métricas y Temporales)
# Resolvemos el error de fechas inconsistentes
df['fecha_hora'] = pd.to_datetime(df['fecha_hora'], dayfirst=True, errors='coerce', format='mixed')

# Convertir métricas a números (float8 y entero4 según el manual técnico)
df['tv_s'] = pd.to_numeric(df['tv_s'], errors='coerce').fillna(0).astype(int)
df['velocidad_km_h'] = pd.to_numeric(df['velocidad_km_h'], errors='coerce').astype(float)
df['longitud_km'] = pd.to_numeric(df['longitud_km'], errors='coerce').astype(float)

# 5. Limpieza Descriptiva
df['nombre_corredor'] = df['nombre_corredor'].str.strip()
df['nombre_dia'] = df['nombre_dia'].str.strip().str.capitalize()

# 6. Eliminar registros que quedaron con fecha inválida
df = df.dropna(subset=['fecha_hora'])

# Guardar el archivo final en tu Drive
df.to_csv('/content/drive/My Drive/velocidad_trafico_LISTO.csv', index=False)

print("¡Proceso completado! Archivo listo para Supabase.")
df.head()
```

```python
import pandas as pd
from google.colab import drive
from google.colab import files
import io

# 1. CONECTAR CON DRIVE
# Esto abrirá una ventana emergente para autorizar el acceso
drive.mount('/content/drive')

# 2. SELECCIONAR EL ARCHIVO
print("Selecciona el archivo de velocidad que quieres arreglar:")
uploaded = files.upload()

# 3. PROCESAR LOS DATOS
nombre_input = list(uploaded.keys())[0]
df_vel = pd.read_csv(io.BytesIO(uploaded[nombre_input]))

# Corregimos las fechas al formato AAAA-MM-DD que pide Supabase
df_vel['fecha_hora'] = pd.to_datetime(df_vel['fecha_hora'], dayfirst=True, errors='coerce').dt.strftime('%Y-%m-%d %H:%M:%S')
df_vel['fecha'] = pd.to_datetime(df_vel['fecha'], dayfirst=True, errors='coerce').dt.strftime('%Y-%m-%d')

# 4. GUARDAR EN DRIVE
# Se guardará con el nombre 'VELOCIDAD_CORREGIDA_FINAL.csv' en la carpeta principal de Drive
ruta_drive = '/content/drive/MyDrive/VELOCIDAD_CORREGIDA_FINAL.csv'
df_vel.to_csv(ruta_drive, index=False)

print(f"✅ ¡ÉXITO! El archivo se guardó en tu Google Drive como: VELOCIDAD_CORREGIDA_FINAL.csv")
```
