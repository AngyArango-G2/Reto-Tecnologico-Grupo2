```python
import pandas as pd
import numpy as np

df = pd.read_csv("velocidad_trafico_limpia.csv", encoding="utf-8-sig")

# 🔥 limpiar nombres de columnas sin romper encoding
df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace("Ã", "A")
    .str.replace("Ñ", "N")
    .str.replace("Í", "I")
    .str.replace("É", "E")
    .str.replace("Á", "A")
    .str.replace("Ó", "O")
    .str.replace("Ú", "U")
    .str.replace(" ", "_")
)

print(df.columns)
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(r"[^\x00-\x7F]+", "", regex=True)  # 🔥 elimina caracteres raros
        .str.strip()
        .str.upper()
    )
```

```python
print(df.columns)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace("Ã\x8d", "I", regex=False)   # Í mal codificada
    .str.replace("Ã\x91", "N", regex=False)   # Ñ mal codificada
    .str.replace("Ã", "A", regex=False)
    .str.replace(" ", "_")
)

print(df.columns)
```
